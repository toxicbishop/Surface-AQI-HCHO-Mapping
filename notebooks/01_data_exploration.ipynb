# 01 · Data Exploration and Network Coverage

This notebook provides an exploratory sanity-check of satellite observation grids, meteorological variables, and ground monitoring station distribution across India.


```
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt

from isro_aqi.synthetic import SyntheticConfig, generate_all
```

### 1. Generate / Load Atmospheric Stack
We initialize the multi-sensor grid (AOD, TROPOMI trace gases, ERA5 meteorology, and MODIS fire detections).


```
cfg = SyntheticConfig(resolution_deg=0.5, n_days=30, n_stations=80)
data = generate_all(cfg)
stack = data["stack"]
stations = data["stations"]
obs = data["observations"]
fires = data["fires"]

print("Grid dimensions:", stack.dims)
print("Predictors:", list(stack.data_vars.keys()))
print(f"Station count: {len(stations)}, Total observations: {len(obs)}")
```

### 2. Station Network Coverage
Visualize ground monitoring station locations over the geographic extent.


```
plt.figure(figsize=(8, 8))
plt.scatter(stations["lon"], stations["lat"], c="crimson", s=30, alpha=0.85, edgecolors="k", label="CPCB Stations")
plt.title("Ground Monitoring Station Coverage (India)")
plt.xlabel("Longitude (°E)")
plt.ylabel("Latitude (°N)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

### 3. Satellite Atmospheric Columns
Display seasonal mean maps of Aerosol Optical Depth (AOD), NO₂, and HCHO.


```
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

stack["aod"].mean(dim="time").plot(ax=axes[0], cmap="YlOrRd")
axes[0].set_title("Mean AOD (Aerosols)")

stack["no2"].mean(dim="time").plot(ax=axes[1], cmap="inferno")
axes[1].set_title("Mean Tropospheric NO₂ Column")

stack["hcho"].mean(dim="time").plot(ax=axes[2], cmap="magma")
axes[2].set_title("Mean Tropospheric HCHO Column")

plt.tight_layout()
plt.show()
```

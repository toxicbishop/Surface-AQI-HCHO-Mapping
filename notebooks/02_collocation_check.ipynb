# 02 · Collocation & Spatial Alignment Check

This notebook verifies spatial and temporal matching between gridded satellite/meteorology fields and point ground monitoring stations.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from isro_aqi.synthetic import SyntheticConfig, generate_all
from isro_aqi.preprocessing.collocate import sample_at_stations, join_targets
```

### 1. Spatio-Temporal Sampling
Extract grid predictors at station coordinates and merge with observed ground measurements.


```
cfg = SyntheticConfig(resolution_deg=0.5, n_days=30, n_stations=60)
data = generate_all(cfg)

sampled = sample_at_stations(data["stack"], data["stations"])
collocated = join_targets(sampled, data["observations"])
print(f"Collocated table shape: {collocated.shape}")
collocated.head()
```

### 2. Predictor vs Target Correlations
Examine the correlation between satellite columnar measurements and ground-level concentrations.


```
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(collocated["aod"], collocated["pm25"], alpha=0.3, color="teal", s=15)
axes[0].set_title("AOD vs Ground PM2.5")
axes[0].set_xlabel("Aerosol Optical Depth (AOD)")
axes[0].set_ylabel("Observed PM2.5 (µg/m³)")
axes[0].grid(True, linestyle="--", alpha=0.5)

axes[1].scatter(collocated["no2"], collocated["no2_obs"], alpha=0.3, color="coral", s=15)
axes[1].set_title("TROPOMI NO₂ Column vs Surface NO₂")
axes[1].set_xlabel("NO₂ Column (mol/m²)")
axes[1].set_ylabel("Observed Surface NO₂ (µg/m³)")
axes[1].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
